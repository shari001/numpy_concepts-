```python
c = np.array([[12,34,124],
             [76,32,198]])
np.savetxt('file.csv',c,delimiter = ',')
```

```python
np.loadtxt('file.csv',delimiter = ',')
```

```python
with_space = ([[12,34,56,43],
              [85,21,243,434]])
np.savetxt('with_space.txt',with_space,delimiter = ' ')

s_data = np.loadtxt('with_space.txt',delimiter = ' ')
print(s_data)
```

```python
import numpy as np

p_data = np.array([[72,43,56,65],
                   [45,64,35,61]])

np.savetxt('with_header.txt', p_data, delimiter = ',',
           header='maths,english,chemistry,physic', fmt='%d')

# Skip the header row
d = np.loadtxt('with_header.txt', delimiter=',')
print(d)
```
